Вам предстоит поработать со списком возрастов посетителей фитнес-клубов age_lst

**Задача 1.** Рассчитайте средний возраст по всему списку age_lst

**Задача 2.** Рассчитайте средний возраст для посетителей, которые моложе 50 лет

**Задача 3.** Оберните предыдущий код в функцию, которая на вход принимает список и пороговое значение

**Задача 4.** Рассчитайте среднее значение для всех элементов списка, за исключением тех, которые лежат вне интервала [Q1-1,5*IQR; Q3+1,5*IQR]

**Задача 5 (со звёздочкой).** Оберните предыдущий код в функцию, которая принимает на вход список и рассчитывает средний возраст, исключая из расчётов выбросы

```python
age_lst = [23, 45, 18, 80, 22, 25, 34, 37, 41, 88, 19, 20, 18, 45, 41, 39, 31, 32, 22, 28, 37]
```

**Задача 1.** Рассчитайте средний возраст по всему списку age_lst

```python
avg_age = sum(age_lst) / len(age_lst)
print(f"Средний возраст по списку составляет {round(avg_age, 2)} лет")
```

**Задача 2.** Рассчитайте средний возраст для посетителей, которые моложе 50 лет

```python
cnt_age = 0
sum_age = 0

for i in age_lst:
    if i < 50:
        sum_age += i
        cnt_age += 1
avg_age2 = sum_age / cnt_age
print(f"Средний возраст по списку для людей моложе 50 лет составляет {round(avg_age2, 2)} лет")
```

**Задача 3.** Оберните предыдущий код в функцию, которая на вход принимает список и пороговое значение

```python
def calc_avg_age(lst, threshold):
    cnt_age = 0
    sum_age = 0
    for i in lst:
        if i < threshold:
            sum_age += i
            cnt_age += 1
    avg_age3 = sum_age / cnt_age
    return print(f"Средний возраст по списку для людей моложе {threshold} лет составляет {round(avg_age3, 2)} лет")
```

```python
calc_avg_age(age_lst, 50)
```

**Задача 4.** Рассчитайте среднее значение для всех элементов списка, за исключением тех, которые лежат вне интервала [Q1-1,5IQR; Q3+1,5IQR]

```python
import numpy as np

q1 = np.percentile(age_lst, 25)
q3 = np.percentile(age_lst, 75)
iqr = q3 - q1

cnt_age = 0
sum_age = 0
for i in age_lst:
    if (i > q1 - 1.5 * iqr) and (i < q3 + 1.5 * iqr):
        sum_age += i
        cnt_age += 1
avg_age4 = sum_age / cnt_age

print(f"Средний возраст по списку за исключением выбросов составляет {round(avg_age4, 2)} лет")
```

**Задача 5 (со звёздочкой).** Оберните предыдущий код в функцию, которая принимает на вход список и рассчитывает средний возраст, исключая из расчётов выбросы

```python
import numpy as np


def calc_avg_age_adv(lst):
    try:
        q1 = np.percentile(lst, 25)
        q3 = np.percentile(lst, 75)
        iqr = q3 - q1
        cnt_age = 0
        sum_age = 0
        for i in lst:
            if (i > q1 - 1.5 * iqr) and (i < q3 + 1.5 * iqr):
                sum_age += i
                cnt_age += 1
        avg_age5 = sum_age / cnt_age
        return print(f"Средний возраст по списку за исключением выбросов составляет {round(avg_age5, 2)} лет")
    except:
        print("Входной список не должен быть пустым или содержать нечисловые значения")
```

```python
new_age_lst = [23, 45, 18, 80, 22, 25, 34, 37, 41, 88, 19, 20, 18, 45, 41, 39, 31, 32, 22, 28, 37, 20,30, 40,20]

calc_avg_age_adv(new_age_lst)
```
